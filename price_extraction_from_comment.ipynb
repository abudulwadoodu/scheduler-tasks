```python
def get_xpath(element):
    tag = element.get("tag", "*")
    
    if element.get("id"):
        return f'//*[@id="{element["id"]}"]'
    
    name = element.get("name", "")
    type_ = element.get("type", "")
    if name and type_:
        return f'//{tag}[@name="{name}" and @type="{type_}"]'
    elif name:
        return f'//{tag}[@name="{name}"]'
    
    class_name = element.get("class_name", "")
    if class_name:
        return f'//{tag}[contains(@class, "{class_name}")]'
    
    return f'//{tag}'
```

```python
def get_xpath_map(sample: dict) -> list:
    xpath_map = []

    for elem in sample["inputs"]:
        tag       = elem.get("tag", "input")
        elem_id   = elem.get("id", "")
        elem_name = elem.get("name", "")
        elem_type = elem.get("type", "").lower()
        elem_label = elem.get("label", "")

        if not elem_id and not elem_name:
            continue

        if elem_id:
            xpath = f"//select[@id='{elem_id}']" if tag == "select" else f"//input[@id='{elem_id}']"
        else:
            xpath = f"//{tag}[@name='{elem_name}']"

        if tag == "select":         action_type = "select"
        elif elem_type == "radio":  action_type = "radio"
        elif elem_type == "checkbox": action_type = "checkbox"
        else:                       action_type = "input"

        xpath_map.append((xpath, elem_label))  # ← tuple, not string

    return xpath_map  # ← list of tuples
```

```python
from lxml import etree

def extract_elements(html: str, xpath_map: list) -> str:
    tree = etree.HTML(html)
    results = []

    for xpath, elem_label in xpath_map:
        els = tree.xpath(xpath)
        if els:
            results.append(f"label name - {elem_label}:\n")
            results.append(etree.tostring(els[0], encoding="unicode"))

    return "\n".join(results)
```

```python
def map_xpath_to_values(parsed_values: dict, xpath_map: list) -> list:
    """
    Maps each valueN to its xpath, element type and action value.
    Returns list of dicts ready for Playwright execution.
    """
    actions = []

    for i, (xpath, elem_type) in enumerate(xpath_map):
        key = f"value{i + 1}"
        value = parsed_values.get(key)

        if value is None:
            continue

        actions.append({
            "xpath": xpath,
            "type": elem_type,
            "value": value
        })

    return actions


```

```python
import re

def parse_comment(comment: str) -> dict:
    values = []

    # --- 1. Frame width & height ---
    # Extract dimensions like "630 x 600mm"
    match = re.search(r"(\d+)\s*x\s*(\d+)", comment)
    if match:
        width, height = match.groups()
    else:
        width, height = "", ""

    values.append(width)   # value1
    values.append(height)  # value2

    # --- 2. Cill (4 radios: No, 85mm, 150mm, 180mm) ---
    # "standard cill" → assume 85mm as sensible default
    cill_selected = "85mm" if "cill" in comment.lower() else None
    cill_options = ["No", "85mm", "150mm", "180mm"]

    for opt in cill_options:
        values.append("check" if opt == cill_selected else None)

    # --- 3. Colour (17 radios) ---
    # Rule: White UPVC → all None (base product)
    colour_options = [
        "White", "Oak Both Sides", "Oak/White", "Rosewood Both Sides",
        "Rosewood/White", "Anthracite Grey Both Sides", "Anthracite Grey/White",
        "Chartwell/White", "Cream Both Sides", "Cream/White",
        "Black-Brown Both Sides", "Black-Brown/White", "Whitegrain Both Sides",
        "Irish Oak Both Sides", "Smooth Anthracite Grey/White", "Agate Grey/White"
    ]

    values.extend([None] * len(colour_options))

    # --- 4. Glass (2 radios: Clear, Obscure) ---
    # Default sensible → Clear
    glass_selected = "Clear"
    glass_options = ["Clear", "Obscure"]

    for opt in glass_options:
        values.append("check" if opt == glass_selected else None)

    # --- 5. Argon glass (2 radios: A Rated, Triple) ---
    argon_selected = "A Rated" if "a energy" in comment.lower() else None
    argon_options = ["A Rated", "A++ Triple Glazed"]

    for opt in argon_options:
        values.append("check" if opt == argon_selected else None)

    # --- 6. Toughened checkbox ---
    values.append("check" if "toughened" in comment.lower() else None)

    # --- 7. Laminated checkbox ---
    values.append("check" if "laminated" in comment.lower() else None)

    # --- 8. Trickle vents (select) ---
    # "trickle vent" → assume 1
    if "trickle vent" in comment.lower():
        trickle_value = "1"
    else:
        trickle_value = "Not Required"

    values.append(trickle_value)

    # --- 9. Fit pack checkbox ---
    values.append("check" if "fit pack" in comment.lower() else None)

    # --- Final dict ---
    return {f"value{i+1}": v for i, v in enumerate(values)}
```

```python
import re


def parse_comment(comment: str) -> dict:
    """
    Learned by cross-referencing comment tokens against HTML option labels:

      attribute188 options are "NNmm" format  → assign the "NNmm" token
      attribute187 options are plain integers  → assign the plain integer token

    Extraction is pattern-based (not option-list matching) so it works
    for any comment value, even ones not present in the dropdown.
    select_option_safe() handles the actual DOM matching + fallback.
    """

  
    # Token: number followed immediately by "mm"
    mm_match = re.search(r"(?<!\d)(\d+(?:\.\d+)?mm)(?!\w)", comment, re.IGNORECASE)

    # Token: plain integer before "x" separator (qty pattern)
    int_match = re.match(r"^\s*(\d+)\s*[xX×]", comment)

    # value1 → attribute188 expects "NNmm" format
    value1 = mm_match.group(1).lower() if mm_match else None

    # value2 → attribute187 expects plain integer
    value2 = int_match.group(1) if int_match else None

    return {
        "value1": value1,  # e.g. "25mm" → select_option_safe matches against "15mm","22mm"...
        "value2": value2,  # e.g. "15"   → select_option_safe matches against "20","25"...
    }


# ── Tests ──────────────────────────────────────────────────────────────────────
if __name__ == "__main__":
    print(parse_comment("15 x 25mm H&V Lag Foil Covered"))
    # → {'value1': '25mm', 'value2': '15'}

    print(parse_comment("30 x 28mm H&V Lag Foil Covered"))
    # → {'value1': '28mm', 'value2': '30'}

    print(parse_comment("H&V Lag Foil Covered"))
    # → {'value1': None, 'value2': None}
```
