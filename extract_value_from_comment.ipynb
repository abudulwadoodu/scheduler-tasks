````python
# function to generate XPath based on element attributes
def get_xpath(element):
    tag = element.get("tag", "*")
    
    if element.get("id"):
        return f'//*[@id="{element["id"]}"]'
    
    name = element.get("name", "")
    type_ = element.get("type", "")
    if name and type_:
        return f'//{tag}[@name="{name}" and @type="{type_}"]'
    elif name:
        return f'//{tag}[@name="{name}"]'
    
    class_name = element.get("class_name", "")
    if class_name:
        return f'//{tag}[contains(@class, "{class_name}")]'
    
    return f'//{tag}'
````

````python
# function to generate a list of (XPath, label) tuples for all input elements
def get_xpath_map(sample: dict) -> list:
    xpath_map = []

    for elem in sample["inputs"]:
        tag       = elem.get("tag", "input")
        elem_id   = elem.get("id", "")
        elem_name = elem.get("name", "")
        elem_type = elem.get("type", "").lower()
        elem_label = elem.get("label", "")

        if not elem_id and not elem_name:
            continue

        if elem_id:
            xpath = f"//select[@id='{elem_id}']" if tag == "select" else f"//input[@id='{elem_id}']"
        else:
            xpath = f"//{tag}[@name='{elem_name}']"

        if tag == "select":         action_type = "select"
        elif elem_type == "radio":  action_type = "radio"
        elif elem_type == "checkbox": action_type = "checkbox"
        else:                       action_type = "input"

        xpath_map.append((xpath, elem_label))  # ← tuple, not string

    return xpath_map  # ← list of tuples
````

````python
from lxml import etree

def extract_elements(html: str, xpath_map: list) -> str:
    tree = etree.HTML(html)
    results = []

    for xpath, elem_label in xpath_map:
        els = tree.xpath(xpath)
        if els:
            results.append(f"label name - {elem_label}\n")
            results.append(etree.tostring(els[0], encoding="unicode"))

    return "\n".join(results)
````

````python
def map_xpath_to_values(parsed_values: dict, xpath_map: list) -> list:
    """
    Maps each valueN to its xpath, element type and action value.
    Returns list of dicts ready for Playwright execution.
    """
    actions = []

    for i, (xpath, elem_type) in enumerate(xpath_map):
        key = f"value{i + 1}"
        value = parsed_values.get(key)

        if value is None:
            continue

        actions.append({
            "xpath": xpath,
            "type": elem_type,
            "value": value
        })

    return actions


````

````python
# pipelagging type 4 comment parser
import re

def parse_comment(comment: str, labels: list) -> dict:
    """
    Parse a comment like 'Armaflex 13mm Class O Insulation Tube 12mm Dia'
    into a dict keyed by label, with the correct <select> option value.

    Pipe Size options  (id=attribute186): 6,10,12,15,20,22,28,35,42,48,54,60,67,76,80,89,93,108,114
    Thickness options  (id=attribute185): 6,9,13,19,25,32
    """

    # ---- Option maps: leading number -> option value attribute ----
    PIPE_SIZE_OPTIONS = {
        "6": "122", "10": "123", "12": "124", "15": "125",
        "20": "127", "22": "128", "28": "129", "35": "130",
        "42": "131", "48": "164", "54": "132", "60": "133",
        "67": "134", "76": "135", "80": "136", "89": "137",
        "93": "138", "108": "139", "114": "140",
    }

    THICKNESS_OPTIONS = {
        "6": "116", "9": "165", "13": "118",
        "19": "119", "25": "120", "32": "121",
    }

    c = comment.lower()

    # ---- Extract all numbers with optional unit context ----
    # Find all (number, trailing_word) pairs e.g. "13mm", "12mm dia"
    tokens = re.findall(r"(\d+)\s*(mm|inch|\")?", c)
    numbers = [t[0] for t in tokens if t[0]]

    pipe_size  = None
    thickness  = None

    # ---- Keyword-first matching ----
    # Look for explicit diameter/pipe-size keywords
    dia_match = re.search(r"(\d+)\s*mm\s*(dia|od|pipe|bore)", c)
    if dia_match:
        pipe_size = dia_match.group(1)

    # Look for explicit thickness keywords
    thick_match = re.search(r"(\d+)\s*mm\s*(thick|wall|insulation|class)", c)
    if thick_match:
        thickness = thick_match.group(1)

    # ---- Fallback: positional assignment ----
    # If keywords didn't resolve both, use order: first number=thickness, second=pipe size
    # (industry convention: "13mm insulation for 12mm pipe")
    if thickness is None or pipe_size is None:
        # Filter to numbers that are plausible for each field
        thickness_candidates = [n for n in numbers if n in THICKNESS_OPTIONS]
        pipe_candidates      = [n for n in numbers if n in PIPE_SIZE_OPTIONS]

        if thickness is None and thickness_candidates:
            thickness = thickness_candidates[0]
        if pipe_size is None:
            # Avoid reusing the number already picked for thickness
            remaining = [n for n in pipe_candidates if n != thickness]
            if remaining:
                pipe_size = remaining[0]

    # ---- Resolve to option values (default to empty string if not found) ----
    pipe_value      = PIPE_SIZE_OPTIONS.get(pipe_size, "")
    thickness_value = THICKNESS_OPTIONS.get(thickness, "")

    # ---- Build result in label order ----
    result = {}
    for label in labels:
        if label == "Insulation to Suit Pipe Size":
            result[label] = pipe_value
        elif label == "Insulation Thickness":
            result[label] = thickness_value
        else:
            result[label] = ""

    return result
````
