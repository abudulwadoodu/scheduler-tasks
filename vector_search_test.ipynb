# Vector Search and Cross-Table Filtering Test

This notebook demonstrates the capability of `VectorDB` to perform similarity searches and filter results using a single `search_term` matched across multiple tables (URI, source, timestamp, and webpage type).

```python
from database.vectordb import VectorDB
import json

vdb = VectorDB()
```

## Basic Similarity Search

Perform a standard vector similarity search without keyword filters.

```python
results = vdb.similarity_search("Baleno", top_k=3, filters={})
print(json.dumps(results, indent=2, default=str))
```

## Cross-Table Search Term Filtering

Search for a term across URL, source name, timestamp, and webpage type. This is now the only filter supported.

```python
# Example: Search for items from 'Amazon' or with 'type_1' webpage type
results = vdb.similarity_search("Show me bikes", top_k=5, filters={"search_term": "Amazon"})
print("Results matching 'Amazon' in any field:")
for r in results:
    print(f"- {r['item_name']} (Dist: {r['distance']:.4f}) | Source: {r['source_name']} | Type: {r['webpage_type']}")
```

## Retrieving All Documents with Metadata

The `get_all_documents` method returns extended metadata and supports the same `search_term` filter.

```python
all_docs = vdb.get_all_documents(filters={"search_term": "Flipkart"})
print(f"Found {len(all_docs)} documents for 'Flipkart':")
for d in all_docs[:3]:
    print(f"- {d['item_name']} | URL: {d['url']} | Timestamp: {d['timestamp']}")
```
