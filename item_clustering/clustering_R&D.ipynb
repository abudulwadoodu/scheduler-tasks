```python
# import
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sentence_transformers import SentenceTransformer
import umap
import hdbscan
import warnings

from training_pipeline import train_clustering_pipeline, save_clustering_model, load_clustering_model, FeatureTransformer

warnings.filterwarnings('ignore')
```

#### Train model


```python
# prepare dataset
source = 'metals4u.co'
sample_data = pd.read_excel("../dataset/materials.xlsx")
sample_data = sample_data[sample_data['Source\n']==source]
sample_data.rename(columns={'Unit\n':'Unit',"Rate £":"Rate"},inplace=True)
train_dataset = sample_data
inference_dataset = sample_data[-4:]
```

```python
    
# Train the pipeline
clustered = train_clustering_pipeline(train_dataset, preference='permissive')
    
# Display results
print("\nSample of clustered data:")
print(clustered['df_clustered'][['Description', 'Unit', 'Rate', 'cluster_type', 'cluster_probability']].head(10))
clustered['df_clustered'].to_excel("cluster_result.xlsx", index=False)
```
