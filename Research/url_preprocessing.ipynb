```python
from urllib.parse import urlparse

def extract_domain(url: str) -> str:
    parsed = urlparse(url)
    return parsed.netloc.lower()
```

```python
import redis

redis_client = redis.Redis(
    host="localhost",
    port=6379,
    decode_responses=True
)
```

```python
URL_STREAMS = [
    "url_stream_1",
    "url_stream_2",
    "url_stream_3",
]

DOMAIN_STREAM_MAP_KEY = "domain_to_stream"
```

```python
def pick_stream(redis_client) -> str:
    idx = redis_client.incr("stream_rr_counter")
    return URL_STREAMS[(idx - 1) % len(URL_STREAMS)]
```

```python
def route_url(url: str, redis_client):
    domain = extract_domain(url)

    # 1. Check if domain already mapped
    stream = redis_client.hget(DOMAIN_STREAM_MAP_KEY, domain)

    if not stream:
        # 2. Assign new stream
        stream = pick_stream(redis_client)

        # 3. Save mapping (sticky)
        redis_client.hset(DOMAIN_STREAM_MAP_KEY, domain, stream)

    # 4. Publish URL to stream
    redis_client.xadd(stream, {"url": url, "domain": domain})

    return stream
```

```python
urls = [
    "https://www.amazon.com/item1",
    "https://www.flipkart.com/item1",
    "https://www.amazon.com/item2",
    "https://www.myntra.com/item1",
    "https://www.flipkart.com/item2",
]

for url in urls:
    stream = route_url(url, redis_client)
    print(f"{url} → {stream}")
```

```python
from datetime import datetime
import redis
import importlib
from urllib.parse import urlparse


def domain_to_module(domain: str):
    domain = domain.replace("www.", "")
    return domain.replace(".", "_")

def load_extractor(domain: str):
    module_name = domain_to_module(domain)
    print("Loading extractor module:", module_name)
    full_path = f"extractors.{module_name}"
    try:
        module = importlib.import_module(full_path)
        return module.extract
    except ModuleNotFoundError:
        print(f"[extractor] No extractor module found for domain: {domain}")
        return None
```
